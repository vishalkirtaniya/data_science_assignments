## 3: Add a column SaleCategory using .apply() with this logic:

- TotalSale >= 50000 → "High"
- TotalSale >= 20000 → "Medium"
- Anything below → "Low"

```python
def logic(sale):
    if sale >= 50000:
        return "High"
    elif sale >= 20000:
        return "Medium"
    else:
        return "Low"

df['SaleCategory'] = df['TotalSale'].apply(logic)
df
```

# Part E — Melt & Pivot

You have this wide-format summary table — create it as a new DataFrame:

```python
summary = pd.DataFrame({
    "SalesRep": ["Raj", "Priya", "Amit"],
    "Q1_Sales": [48000, 12500, 36000],
    "Q2_Sales": [32000, 54000, 22000],
    "Q3_Sales": [61000, 29000, 47000]
})
```

## 1: Use melt() to convert it to long format with columns SalesRep, Quarter, Sales

```python
melted = summary.melt(id_vars='SalesRep', var_name="Quarter", value_name="Sales")
melted
```

## 2: Use pivot() to convert the melted DataFrame back to wide format

```python
wide = melted.pivot(index="SalesRep", columns="Quarter", values="Sales")
wide
```

## 3: Confirm both DataFrames have the same values — what do you notice about the column order?

```python
'''
    the melted dataframe contains more more rows and repeated data,
    and pivot dataframe has only 3 rows but contains the same amount of input as melt dataframe
'''
```
